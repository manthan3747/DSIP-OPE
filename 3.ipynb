```python
import librosa
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import correlate

# Load the three audio files
song1, sr1 = librosa.load("OPE3audio(tokyo drift).mp3", sr=None, mono=True)
song2, sr2 = librosa.load("OPE3audio(karaoke).mp3", sr=None, mono=True)
song3, sr3 = librosa.load("OPE3audio(Danza kuduro).mp3", sr=None, mono=True)

# Make all signals the same length
min_length = min(len(song1), len(song2), len(song3))

song1 = song1[:min_length]
song2 = song2[:min_length]
song3 = song3[:min_length]

# Normalize the signals
song1 = song1 / np.max(np.abs(song1))
song2 = song2 / np.max(np.abs(song2))
song3 = song3 / np.max(np.abs(song3))

# Calculate normalized correlation
def find_correlation(signal1, signal2):

    corr = correlate(signal1, signal2, mode="full")

    corr = corr / (np.sqrt(np.sum(signal1 ** 2)) * np.sqrt(np.sum(signal2 ** 2)))

    return corr

# Find correlation between Tokyo Drift and Karaoke
corr1 = find_correlation(song1, song2)

# Find correlation between Tokyo Drift and Danza Kuduro
corr2 = find_correlation(song1, song3)

# Find correlation between Karaoke and Danza Kuduro
corr3 = find_correlation(song2, song3)

# Find maximum correlation values
value1 = np.max(corr1)
value2 = np.max(corr2)
value3 = np.max(corr3)

# Print the results
print("--------------------------------------")
print("Maximum Correlation Values")
print("--------------------------------------")
print("Tokyo Drift vs Karaoke      :", value1)
print("Tokyo Drift vs Danza Kuduro :", value2)
print("Karaoke vs Danza Kuduro     :", value3)

# Plot first audio signal
plt.figure(figsize=(10,4))
plt.plot(song1)
plt.title("Tokyo Drift")
plt.xlabel("Samples")
plt.ylabel("Amplitude")
plt.grid(True)

# Plot second audio signal
plt.figure(figsize=(10,4))
plt.plot(song2)
plt.title("Karaoke")
plt.xlabel("Samples")
plt.ylabel("Amplitude")
plt.grid(True)

# Plot third audio signal
plt.figure(figsize=(10,4))
plt.plot(song3)
plt.title("Danza Kuduro")
plt.xlabel("Samples")
plt.ylabel("Amplitude")
plt.grid(True)

# Plot correlation between Tokyo Drift and Karaoke
plt.figure(figsize=(10,4))
plt.plot(corr1)
plt.title("Correlation : Tokyo Drift vs Karaoke")
plt.xlabel("Lag")
plt.ylabel("Correlation")
plt.grid(True)

# Plot correlation between Tokyo Drift and Danza Kuduro
plt.figure(figsize=(10,4))
plt.plot(corr2)
plt.title("Correlation : Tokyo Drift vs Danza Kuduro")
plt.xlabel("Lag")
plt.ylabel("Correlation")
plt.grid(True)

# Plot correlation between Karaoke and Danza Kuduro
plt.figure(figsize=(10,4))
plt.plot(corr3)
plt.title("Correlation : Karaoke vs Danza Kuduro")
plt.xlabel("Lag")
plt.ylabel("Correlation")
plt.grid(True)

plt.show()
```

```python
import librosa
import numpy as np
import matplotlib.pyplot as plt

# Load the three audio files
song1, sr1 = librosa.load("OPE3audio(tokyo drift).mp3", sr=None, mono=True)
song2, sr2 = librosa.load("OPE3audio(karaoke).mp3", sr=None, mono=True)
song3, sr3 = librosa.load("OPE3audio(Danza kuduro).mp3", sr=None, mono=True)

# Make all signals the same length
min_length = min(len(song1), len(song2), len(song3))

song1 = song1[:min_length]
song2 = song2[:min_length]
song3 = song3[:min_length]

# Function to calculate phase correlation
def phase_correlation(signal1, signal2):

    fft1 = np.fft.fft(signal1)
    fft2 = np.fft.fft(signal2)

    cross_power = fft1 * np.conjugate(fft2)

    cross_power = cross_power / (np.abs(cross_power) + 1e-10)

    phase = np.fft.ifft(cross_power)

    return np.abs(phase)

# Calculate phase correlation
phase1 = phase_correlation(song1, song2)
phase2 = phase_correlation(song1, song3)
phase3 = phase_correlation(song2, song3)

# Print maximum phase correlation values
print("Phase Correlation Results")
print("--------------------------------")
print("Tokyo Drift vs Karaoke      :", np.max(phase1))
print("Tokyo Drift vs Danza Kuduro :", np.max(phase2))
print("Karaoke vs Danza Kuduro     :", np.max(phase3))

# Plot graphs
plt.figure(figsize=(10,4))
plt.plot(phase1)
plt.title("Phase Correlation : Tokyo Drift vs Karaoke")
plt.xlabel("Samples")
plt.ylabel("Correlation")
plt.grid(True)

plt.figure(figsize=(10,4))
plt.plot(phase2)
plt.title("Phase Correlation : Tokyo Drift vs Danza Kuduro")
plt.xlabel("Samples")
plt.ylabel("Correlation")
plt.grid(True)

plt.figure(figsize=(10,4))
plt.plot(phase3)
plt.title("Phase Correlation : Karaoke vs Danza Kuduro")
plt.xlabel("Samples")
plt.ylabel("Correlation")
plt.grid(True)

plt.show()
```
