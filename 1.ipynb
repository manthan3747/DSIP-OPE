```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-3, 7)

x = []

for i in n:
    if i % 2 == 0:
        x.append(3)
    else:
        x.append(2)

plt.stem(n, x)
plt.grid(True)
plt.title("Periodic sequence (2,3,2,3,...)")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(0, 6)

x = [1, 2, 3, 3, 2, 1]

plt.step(n, x, where='post')
plt.grid(True)
plt.title("Staircase Signal")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-2, 3)

x = []

for i in n:
    if i == -2:
        x.append(-2)
    elif i == -1:
        x.append(-2)
    elif i == 0:
        x.append(0)
    elif i == 1:
        x.append(2)
    else:
        x.append(2)

plt.stem(n, x)
plt.grid(True)
plt.title("Piecewise Signal")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-2, 11)

u1 = []
u2 = []
u3 = []

for i in n:

    if i >= 0:
        u1.append(1)
    else:
        u1.append(0)

    if i >= 3:
        u2.append(1)
    else:
        u2.append(0)

    if i >= 7:
        u3.append(1)
    else:
        u3.append(0)

x = []

for i in range(len(n)):
    value = u1[i] - u2[i] - 5*u3[i]
    x.append(value)

plt.stem(n, x)
plt.grid(True)
plt.title("x[n]=u(n)−u(n−3)−5u(n−7)")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

n = np.arange(-3, 4)

x = []

for i in n:

    if i == -1:
        x.append(5)

    elif i == 0:
        x.append(1)

    elif i == 1:
        x.append(3)

    else:
        x.append(0)

plt.stem(n, x)
plt.grid(True)
plt.title("x[n]=δ(n)+3δ(n−1)+5δ(n+1)")
plt.xlabel("n")
plt.ylabel("x[n]")
plt.show()
```
